```python
import requests


def CallAPI(name, description):
    api_url = 'http://localhost:8080/api/admin/skincare'

    headers = {
        "token": " YOUR_API_KEY",  # Replace YOUR_API_KEY with your actual API key
    }

    form_data = {
        "name": name,
        "description": description
    }

    # Make the API call
    response = requests.post(api_url, headers=headers, json=form_data)

    # Check for errors and return the response
    if response.status_code == 200:
        return response.json()
    else:
        return {"error": response.status_code, "message": response.text}
```
