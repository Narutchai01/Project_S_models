```python
import pandas as pd



df = pd.read_csv("D:\\Work\\Project_S_models\\notebooks\\recommend\\datasheet.csv")

unwanted_types = [
    "Face Makeup", "Bath and Body", "Other Hair Care", "Shampoo", "Fragrance", 
    "Hand Care", "Condi9oner", "Lip Moisturizer", "Eye Makeup", "tanning", 
    "nan (เสิชละ้ไม่เจอไร)", "Cheek Makeup", "Nail Care", "Lip Mask", 
    "Other", "Other Haircare", "Tool", "False Eyelash"
]

df = df[~df['type'].isin(unwanted_types)]
df
```

```python
# Update type to 'Pads' if name contains 'Pads' or 'pad' and type is 'Facial Treatment'

newdf = df.copy()

newdf.loc[(newdf['name'].str.contains('Pads', case=False)) & (
    newdf['type'] == 'Facial Treatment'), 'type'] = 'Pads'

# Update type to 'Cream' if name contains 'Cream' and type is one of the specified types
newdf.loc[(newdf['name'].str.contains('Cream', case=False)) &
          (newdf['type'].isin(['Night Moisturizer', 'Day Moisturizer', 'Face Cream', 'General Moisturizer'])), 'type'] = 'Cream'

newdf.loc[(newdf['name'].str.contains('Lotion', case=False)) &
          (newdf['type'].isin(['Night Moisturizer', 'Day Moisturizer', 'Face Cream', 'General Moisturizer'])), 'type'] = 'Lotion'

newdf.loc[(newdf['name'].str.contains('Serum', case=False)) &
          (newdf['type'].isin(['Facial Treatment'])), 'type'] = 'Serum'

newdf.loc[(newdf['name'].str.contains('Serum', case=False)) &
          (newdf['type'].isin(['Facial Treatment'])), 'type'] = 'Serum'

newdf.loc[(newdf['name'].str.contains('Mist', case=False)) &
          (newdf['type'].isin(['Facial Treatment'])), 'type'] = 'Mist'

newdf.loc[(newdf['name'].str.contains('Spray', case=False)) &
          (newdf['type'].isin(['Facial Treatment'])), 'type'] = 'Mist'

newdf.loc[(newdf['name'].str.contains('Gel', case=False)) &
          (newdf['type'].isin(['Facial Treatment'])), 'type'] = 'Gel'

newdf.loc[(newdf['name'].str.contains('Emulsion', case=False)) &
          (newdf['type'].isin(['Emulsion','Facial Treatment'])), 'type'] = 'Emulsion'

newdf.loc[(newdf['name'].str.contains('Facial', case=False)) &
         (newdf['type'].isin(['Oil'])), 'type'] = 'Oil'

newdf.loc[(newdf['name'].str.contains('Face', case=False)) &
         (newdf['type'].isin(['Oil'])), 'type'] = 'Oil'

newdf.loc[(newdf['name'].str.contains('Essence', case=False)) &
          (newdf['type'].isin(['Essence','Facial Treatment'])), 'type'] = 'Essence'
```
